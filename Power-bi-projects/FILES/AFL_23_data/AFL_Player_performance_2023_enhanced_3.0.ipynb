## Outliner removal, decide which predictor we want to predict

```python
def outliners(df,ft):
    Q1 = df[ft].quantile(0.25)
    Q3 = df[ft].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5*IQR
    upper = Q3 + 1.5*IQR
    ls = df.index[(df[ft] < lower)|(df[ft] > upper)]
    return ls
```
